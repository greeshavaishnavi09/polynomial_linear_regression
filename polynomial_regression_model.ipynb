## Polynomial Regression Mode


```python
# Step 1: Import required lib 

import pandas as pd
import matplotlib.pyplot as plt 
import seaborn as sns 
```

```python
# Step 2: Load the datasets

df = pd.read_csv("polynomial_dataset.csv")
```

```python
# Step 3: Data Expoloration 

df.head()
```

```python
df.info()
```

```python
# Step 3: EDA

plt.scatter(df['Temperature (°C)'],df['Ice Cream Sales (units)'])
plt.title('Plot the data')
plt.xlabel("Temperature (°C)")
plt.ylabel("Ice Cream Sales (units)")
plt.show()
```

```python
# Step 5: Divide the into X and Y

X = df['Temperature (°C)'].values.reshape(-1,1)
y= df['Ice Cream Sales (units)']
```

```python
# polynomial regression 

from sklearn.preprocessing import PolynomialFeatures

# create the instance 
poly = PolynomialFeatures(degree=2)

# fit and transform the 
x_poly = poly.fit_transform(X)
```

```python
x_poly
```

```python
# Step 6: Model building 

from sklearn.linear_model import LinearRegression

# create the Instance of model 
LR = LinearRegression()

# Train the model on the X_poly 
model = LR.fit(x_poly,y)

# model prediction 
y_pred = model.predict(x_poly)

```

```python
# model evualtion metrics 

from sklearn.metrics import  mean_absolute_error,mean_squared_error,r2_score

# find the MSE
MSE = mean_squared_error(y,y_pred)

# find the MAE
MAE = mean_absolute_error(y,y_pred)

#find the r2 score
r2_score =  r2_score(y,y_pred)


# print all metrics 
print(f" MSE : {MSE}")
print(f" MAE : {MAE}")
print(f" r2_score : {r2_score}")

```
